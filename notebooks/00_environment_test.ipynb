```python
# Kiểm tra phiên bản Python và đường dẫn thực thi
import sys
print(f"Phiên bản Python: {sys.version}")
print(f"Trình thực thi:    {sys.executable}")
```

```python
# Nạp toàn bộ thư viện cốt lõi của đồ án
import pandas as pd
import numpy as np
import pyarrow as pa
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import statsmodels
import sklearn

print("Tất cả thư viện cốt lõi đã được nạp thành công.")
```

```python
# Thống kê bảng phiên bản các thư viện được cài đặt
libraries = {
    "pandas":       pd.__version__,
    "numpy":        np.__version__,
    "pyarrow":      pa.__version__,
    "matplotlib":   matplotlib.__version__,
    "seaborn":      sns.__version__,
    "scipy":        scipy.__version__,
    "statsmodels":  statsmodels.__version__,
    "scikit-learn": sklearn.__version__,
}

version_df = pd.DataFrame(
    list(libraries.items()),
    columns=["Thư viện", "Phiên bản"],
)
print(version_df.to_string(index=False))
```

```python
# Smoke test 1: Kiểm thử chu trình ghi và đọc tệp Parquet (Snappy)
import tempfile, os

df_test = pd.DataFrame({
    "timestamp": pd.date_range("2024-01-01", periods=5, freq="h"),
    "pm25":      [35.2, 42.1, 38.7, 55.0, 47.3],
})

tmp_path = os.path.join(tempfile.gettempdir(), "env_test.parquet")
df_test.to_parquet(tmp_path, engine="pyarrow", compression="snappy")
df_reload = pd.read_parquet(tmp_path, engine="pyarrow")

assert df_test.shape == df_reload.shape, "Lỗi: Kích thước dữ liệu Parquet không khớp!"
assert (df_test["pm25"] == df_reload["pm25"]).all(), "Lỗi: Dữ liệu Parquet không đồng nhất!"

os.remove(tmp_path)
print("Kiểm thử Parquet round-trip: ĐẠT (PASSED)")
```

```python
# Smoke test 2: Kiểm thử khả năng render biểu đồ hướng đối tượng Matplotlib
fig, ax = plt.subplots(figsize=(4, 2))
ax.plot([1, 2, 3], [1, 4, 9], marker="o")
ax.set_title("Kiểm thử Matplotlib render")
plt.close(fig)
print("Kiểm thử Matplotlib render: ĐẠT (PASSED)")
```

```python
# Smoke test 3: Kiểm thử Pipeline và RobustScaler của Scikit-Learn
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler

pipe = Pipeline([("scaler", RobustScaler())])
X = np.array([[1.0], [2.0], [3.0], [100.0]])
X_scaled = pipe.fit_transform(X)

assert X_scaled.shape == X.shape, "Lỗi: Kích thước đầu ra của Pipeline không khớp!"
print("Kiểm thử Scikit-Learn Pipeline: ĐẠT (PASSED)")
```
